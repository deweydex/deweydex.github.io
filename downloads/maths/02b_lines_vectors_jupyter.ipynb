## Part A: Which Dataset Is Most Linear?

Below are three datasets showing the relationship between "Hours Studied" and "Exam Score". 
Examine each dataset and decide which is best represented by a linear model.

```python
# Three datasets
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
scores_a = np.array([45, 52, 58, 61, 67, 73, 78, 84])  # Linear
scores_b = np.array([50, 55, 70, 60, 85, 65, 90, 75])  # Scattered
scores_c = np.array([40, 45, 55, 70, 82, 88, 92, 94])  # Curved

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].scatter(hours, scores_a, c='steelblue', s=80)
axes[0].set_xlabel('Hours Studied')
axes[0].set_ylabel('Exam Score')
axes[0].set_title('Dataset A')
axes[0].set_xlim(0, 9)
axes[0].set_ylim(30, 100)
axes[0].grid(True, alpha=0.3)

axes[1].scatter(hours, scores_b, c='coral', s=80)
axes[1].set_xlabel('Hours Studied')
axes[1].set_ylabel('Exam Score')
axes[1].set_title('Dataset B')
axes[1].set_xlim(0, 9)
axes[1].set_ylim(30, 100)
axes[1].grid(True, alpha=0.3)

axes[2].scatter(hours, scores_c, c='seagreen', s=80)
axes[2].set_xlabel('Hours Studied')
axes[2].set_ylabel('Exam Score')
axes[2].set_title('Dataset C')
axes[2].set_xlim(0, 9)
axes[2].set_ylim(30, 100)
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Try Fitting a Line to Dataset A

Use the sliders below to adjust the slope and intercept of a line. Try to make it fit Dataset A as closely as possible.

```python
def fit_line_to_data(slope=5.5, intercept=40):
    fig, ax = plt.subplots(figsize=(8, 5))
    ax.scatter(hours, scores_a, c='steelblue', s=80, label='Data', zorder=5)
    
    x_line = np.linspace(0, 9, 100)
    y_line = slope * x_line + intercept
    ax.plot(x_line, y_line, 'r-', linewidth=2, label=f'y = {slope}x + {intercept}')
    
    # Calculate residuals
    predicted = slope * hours + intercept
    residuals = scores_a - predicted
    sse = np.sum(residuals**2)
    
    ax.set_xlabel('Hours Studied')
    ax.set_ylabel('Exam Score')
    ax.set_title(f'Dataset A with Your Line (Sum of Squared Errors: {sse:.1f})')
    ax.set_xlim(0, 9)
    ax.set_ylim(30, 100)
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.show()

interact(fit_line_to_data, 
         slope=widgets.FloatSlider(min=0, max=10, step=0.5, value=5, description='Slope (m)'),
         intercept=widgets.FloatSlider(min=30, max=60, step=1, value=40, description='Intercept (b)'));
```

## Part B: Positive, Negative, or Zero Slope?

Adjust the slope slider and observe how the line changes.

```python
def show_slope(slope=1):
    fig, ax = plt.subplots(figsize=(8, 5))
    
    x = np.linspace(-5, 5, 100)
    y = slope * x
    
    ax.plot(x, y, 'b-', linewidth=2)
    ax.axhline(y=0, color='k', linewidth=0.5)
    ax.axvline(x=0, color='k', linewidth=0.5)
    ax.set_xlim(-5, 5)
    ax.set_ylim(-8, 8)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    
    if slope > 0:
        desc = "POSITIVE slope - line goes UP from left to right"
    elif slope < 0:
        desc = "NEGATIVE slope - line goes DOWN from left to right"
    else:
        desc = "ZERO slope - line is HORIZONTAL"
    
    ax.set_title(f'Slope = {slope}: {desc}')
    ax.grid(True, alpha=0.3)
    plt.show()

interact(show_slope, slope=widgets.FloatSlider(min=-3, max=3, step=0.25, value=1, description='Slope'));
```

## Part C: Eyeballing the Line of Best Fit

Here's a scatter plot. Use the sliders to find the line of best fit by minimizing the residuals.

```python
x_data = np.array([1, 2, 3, 4, 5, 6, 7])
y_data = np.array([2, 3, 5, 4, 6, 7, 8])

def fit_with_residuals(slope=1.0, intercept=1.0, show_residuals=True):
    fig, ax = plt.subplots(figsize=(8, 6))
    
    x_line = np.linspace(0, 8, 100)
    y_line = slope * x_line + intercept
    ax.plot(x_line, y_line, 'r-', linewidth=2, label=f'y = {slope:.2f}x + {intercept:.1f}')
    
    predicted = slope * x_data + intercept
    residuals = y_data - predicted
    
    if show_residuals:
        for i in range(len(x_data)):
            ax.plot([x_data[i], x_data[i]], [y_data[i], predicted[i]], 'g-', linewidth=1.5, alpha=0.7)
    
    ax.scatter(x_data, y_data, c='steelblue', s=100, zorder=5, label='Data')
    
    sse = np.sum(residuals**2)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title(f'Sum of Squared Residuals: {sse:.2f}')
    ax.set_xlim(0, 8)
    ax.set_ylim(0, 10)
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.show()

interact(fit_with_residuals,
         slope=widgets.FloatSlider(min=0, max=2, step=0.05, value=1, description='Slope'),
         intercept=widgets.FloatSlider(min=0, max=3, step=0.1, value=1, description='Intercept'),
         show_residuals=widgets.Checkbox(value=True, description='Show residuals'));
```

## Part D: Slope on Curves — Tangent Line Intuition

> **Key Idea:** On a straight line, the slope is the same everywhere. On a curve, the slope *changes* as you move along it. At any point, we can draw a **tangent line** that shows the slope at that exact point.

### Parabola: y = x²

```python
def parabola_tangent(x_point=1.0):
    fig, ax = plt.subplots(figsize=(9, 6))
    
    x_curve = np.linspace(-3.5, 3.5, 200)
    y_curve = x_curve ** 2
    ax.plot(x_curve, y_curve, 'b-', linewidth=2, label='y = x²')
    
    px, py = x_point, x_point ** 2
    slope = 2 * px
    tangent_x = np.linspace(px - 2, px + 2, 100)
    tangent_y = slope * (tangent_x - px) + py
    
    ax.plot(tangent_x, tangent_y, 'r-', linewidth=2, label=f'Tangent (slope = {slope:.2f})')
    ax.scatter([px], [py], c='red', s=120, zorder=5)
    
    ax.axhline(y=0, color='k', linewidth=0.5)
    ax.axvline(x=0, color='k', linewidth=0.5)
    ax.set_xlim(-4, 4)
    ax.set_ylim(-2, 12)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    
    if slope > 0.1:
        desc = "Positive slope - curve going UP"
    elif slope < -0.1:
        desc = "Negative slope - curve going DOWN"
    else:
        desc = "Zero slope - curve is FLAT here"
    
    ax.set_title(f'At x = {px:.1f}: {desc}')
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.show()

interact(parabola_tangent, x_point=widgets.FloatSlider(min=-3, max=3, step=0.1, value=1, description='x-coord'));
```

**Questions:**
1. At what x-value is the slope zero?
2. For x > 0, is the slope positive or negative?
3. For x < 0, is the slope positive or negative?

### Exponential Growth: y = 2^x

```python
def exponential_tangent(x_point=1.0):
    fig, ax = plt.subplots(figsize=(9, 6))
    
    x_curve = np.linspace(-2.5, 3.5, 200)
    y_curve = 2 ** x_curve
    ax.plot(x_curve, y_curve, 'b-', linewidth=2, label='y = 2^x')
    
    px = x_point
    py = 2 ** px
    slope = py * np.log(2)
    tangent_x = np.linspace(px - 1.5, px + 1.5, 100)
    tangent_y = slope * (tangent_x - px) + py
    
    ax.plot(tangent_x, tangent_y, 'r-', linewidth=2, label=f'Tangent (slope = {slope:.2f})')
    ax.scatter([px], [py], c='red', s=120, zorder=5)
    
    ax.axhline(y=0, color='k', linewidth=0.5)
    ax.axvline(x=0, color='k', linewidth=0.5)
    ax.set_xlim(-3, 4)
    ax.set_ylim(-1, 12)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title(f'Exponential y = 2^x | At x = {px:.1f}, slope = {slope:.2f}')
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.show()

interact(exponential_tangent, x_point=widgets.FloatSlider(min=-2, max=3, step=0.1, value=1, description='x-coord'));
```

**Questions:**
1. Is the slope ever negative?
2. Is the slope ever zero?
3. As x increases, does the slope get steeper or flatter?

### Sine Wave: y = sin(x)

```python
def sine_tangent(x_point=0.0):
    fig, ax = plt.subplots(figsize=(10, 5))
    
    x_curve = np.linspace(-0.5, 7, 300)
    y_curve = np.sin(x_curve)
    ax.plot(x_curve, y_curve, 'b-', linewidth=2, label='y = sin(x)')
    
    px = x_point
    py = np.sin(px)
    slope = np.cos(px)
    tangent_x = np.linspace(px - 1, px + 1, 100)
    tangent_y = slope * (tangent_x - px) + py
    
    ax.plot(tangent_x, tangent_y, 'r-', linewidth=2, label=f'Tangent (slope = {slope:.2f})')
    ax.scatter([px], [py], c='red', s=120, zorder=5)
    
    ax.axhline(y=0, color='k', linewidth=0.5)
    ax.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
    ax.set_xticklabels(['0', 'pi/2', 'pi', '3pi/2', '2pi'])
    ax.set_xlim(-0.5, 7)
    ax.set_ylim(-2, 2)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    
    if slope > 0.1:
        desc = "going UP"
    elif slope < -0.1:
        desc = "going DOWN"
    else:
        desc = "FLAT (peak or trough)"
    
    ax.set_title(f'Sine Wave | At x = {px:.2f}, slope = {slope:.2f} - {desc}')
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.show()

interact(sine_tangent, x_point=widgets.FloatSlider(min=0, max=6.28, step=0.1, value=0, description='x-coord'));
```

**Questions:**
1. At x = 0, is the tangent going up, down, or flat?
2. At x = pi/2 (the peak), what is the slope?
3. Where is the slope most positive? Most negative?

## Part E: Model Training Simulation

```python
def training_curve(learning_rate=0.5):
    fig, ax = plt.subplots(figsize=(9, 5))
    
    epochs = np.arange(0, 101)
    accuracy = 1 - np.exp(-learning_rate * epochs / 20)
    
    ax.plot(epochs, accuracy, 'b-', linewidth=2)
    ax.scatter(epochs[::10], accuracy[::10], c='steelblue', s=60, zorder=5)
    ax.axhline(y=1.0, color='gray', linestyle='--', alpha=0.5, label='Maximum (1.0)')
    
    ax.set_xlabel('Training Epochs')
    ax.set_ylabel('Model Accuracy')
    ax.set_title(f'Model Training (Learning Rate = {learning_rate})')
    ax.set_xlim(0, 100)
    ax.set_ylim(0, 1.1)
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.show()

interact(training_curve, learning_rate=widgets.FloatSlider(min=0.1, max=2.0, step=0.1, value=0.5, description='Learn Rate'));
```

**Questions:**
1. Where is improvement fastest?
2. Where does improvement slow down (diminishing returns)?
3. Will accuracy ever reach 1.0?

## Part F: Correlation Explorer

```python
def correlation_explorer(target_r=0.8, noise=0.5):
    np.random.seed(42)
    n = 50
    x = np.linspace(0, 10, n)
    
    if abs(target_r) > 0.01:
        y = target_r * x + np.random.normal(0, noise * 2, n)
    else:
        y = np.random.normal(5, noise * 2, n)
    
    actual_r = np.corrcoef(x, y)[0, 1]
    
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.scatter(x, y, c='steelblue', s=60, alpha=0.7)
    
    slope, intercept = np.polyfit(x, y, 1)
    ax.plot(x, slope * x + intercept, 'r-', linewidth=2, alpha=0.7, label=f'Best fit')
    
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title(f'Target r = {target_r:.2f} | Actual r = {actual_r:.2f}')
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.show()

interact(correlation_explorer,
         target_r=widgets.FloatSlider(min=-1, max=1, step=0.05, value=0.8, description='Target r'),
         noise=widgets.FloatSlider(min=0, max=2, step=0.1, value=0.5, description='Noise'));
```

**Questions:**
1. What does r = 1 look like? r = -1? r = 0?
2. What does increasing noise do to correlation?
3. With r = 0.8, is the relationship strong or weak?

## Summary

**Key Takeaways:**

1. **Linear data** shows a consistent pattern; curved or scattered data doesn't fit a line well
2. **Slope** tells us the rate of change — positive (increasing), negative (decreasing), or zero (flat)
3. **Residuals** measure how far our predictions are from reality; we minimize squared residuals
4. **On curves**, the slope changes from point to point — the tangent line shows the instantaneous slope
5. **Correlation (r)** measures the strength of a linear relationship, from -1 to +1

These ideas connect directly to machine learning: linear regression finds the best-fit line, gradient descent follows the slope to minimize error, and understanding curves helps us choose appropriate models.
